```python
## 10. Statistical Analysis
### T-Test: Tenure vs Churn
```

```python
from scipy.stats import ttest_ind

churned_tenure = df.loc[df["Churn"] == "Yes", "tenure"]
retained_tenure = df.loc[df["Churn"] == "No", "tenure"]

t_stat, p_value = ttest_ind(
    churned_tenure,
    retained_tenure,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
### Chi-Square Test: Contract Type vs Churn
```

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df["Contract"], df["Churn"])

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square:", chi2)
print("p-value:", p_value)
```

### Statistical Test: Contract Type vs Churn

The chi-square test produced a p-value far below 0.05, indicating a statistically significant association between contract type and customer churn. This supports the finding that customers on month-to-month contracts are substantially more likely to churn than customers on longer-term contracts.

```python
## 11. Statistical Summary
### Comparison of Churned and Retained Customers
```

```python
df.groupby("Churn")[["tenure", "MonthlyCharges", "TotalCharges"]].agg(
    ["mean", "median", "std"]
)
```

### Statistical Summary

Churned customers have substantially lower tenure than retained customers, with an average tenure of 17.98 months compared with 37.65 months. They also have higher average monthly charges (74.44 vs 61.31), but lower average total charges (1531.80 vs 2555.34), which is consistent with their shorter customer lifespan.

## Final Business Insights & Conclusion

### Key Insights

1. **Overall churn rate:** Approximately **26.58%** of customers have churned.

2. **Tenure is strongly associated with churn:** Churned customers have a much shorter average tenure (**17.98 months**) compared with retained customers (**37.65 months**). The t-test also confirms that this difference is statistically significant.

3. **Monthly charges:** Churned customers have higher average monthly charges (**74.44**) than retained customers (**61.31**).

4. **Contract type:** Month-to-month customers have the highest churn rate (**42.71%**), while two-year contract customers have the lowest (**2.85%**). The chi-square test confirms a statistically significant association between contract type and churn.

5. **Internet service:** Customers using fiber optic service show the highest churn rate (**41.89%**), compared with DSL (**19.00%**) and customers without internet service (**7.43%**).

6. **Payment method:** Customers using electronic checks have the highest churn rate (**45.29%**), considerably higher than the other payment methods.

7. **Total charges:** Churned customers have lower average total charges (**1531.80**) than retained customers (**2555.34**), which is consistent with their shorter tenure.

### Conclusion

The analysis shows that customer churn is strongly associated with factors such as **shorter tenure, higher monthly charges, month-to-month contracts, fiber optic internet service, and electronic check payments**.

These findings can help the business identify higher-risk customer segments and develop targeted retention strategies. However, these relationships represent **associations rather than proof of causation**.
